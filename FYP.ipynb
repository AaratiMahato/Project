 To Set up the directory with documents

```python
from google.colab import drive
import os

# Path to the folder containing your text documents
docs_path = "/content/drive/MyDrive/Short Book Excerpts"

# List all files in the folder
documents = []
filenames = []

for filename in os.listdir(docs_path):
    if filename.endswith(".txt"):
        filenames.append(filename)
        with open(os.path.join(docs_path, filename), 'r') as file:
            documents.append(file.read())


```

Tokenization and Preprocessing

```python
import re
from collections import defaultdict

# Tokenization and normalization function
def tokenize(text):
    text = text.lower()
    # Remove punctuation and split into words
    tokens = re.findall(r'\b\w+\b', text)
    return tokens

# Preprocess the loaded documents
tokenized_docs = [tokenize(doc) for doc in documents]
```

Build Inverted Index

```python
# Create an inverted index
inverted_index = defaultdict(set)

for doc_id, tokens in enumerate(tokenized_docs):
    for token in tokens:
        inverted_index[token].add(doc_id)

# Display the inverted index
for term, doc_ids in inverted_index.items():
    print(f"{term}: {sorted(list(doc_ids))}")
```

Boolean Retrieval Function

```python
# Boolean Retrieval function
def boolean_retrieval(query):
    query_terms = query.lower().split()
    result = None

    if 'and' in query_terms:
        term1 = query_terms[0]
        term2 = query_terms[2]
        result = inverted_index[term1] & inverted_index[term2]
    elif 'or' in query_terms:
        term1 = query_terms[0]
        term2 = query_terms[2]
        result = inverted_index[term1] | inverted_index[term2]
    elif 'not' in query_terms:
        term = query_terms[1]
        all_docs = set(range(len(documents)))
        result = all_docs - inverted_index[term]
    else:
        result = inverted_index[query_terms[0]]

    return sorted(list(result))

# Example: Read queries from a file
queries_path = "/content/drive/MyDrive/Query/query.txt"

# Load queries from file (each line is a query)
queries = []
with open(queries_path, 'r') as file:
    queries = [line.strip() for line in file.readlines()]

# Process and display results for each query
for query in queries:
    print(f"Results for query '{query}': {boolean_retrieval(query)}")
```
